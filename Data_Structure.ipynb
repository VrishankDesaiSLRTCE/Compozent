```python
def students(n):
    result = {}
    for _ in range(n):
        roll_no = int(input("Roll No: "))
        std_name = input("Student Name: ")
        marks = int(input("Marks: "))
        result[roll_no] = [std_name, marks]
    print(result)

n = int(input("Enter No of Students:"))
students(n)
```
